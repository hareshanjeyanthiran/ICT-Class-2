```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv('Weather Dataset.csv')
df.head()
```

```python
df.dtypes
```

```python
df.isnull().sum()
```

```python
Temperature_mean = df['Temperature (C)'].mean()
print("Mean of Temperature:", Temperature_mean)
```

```python
var_temp = np.var(df['Temperature (C)'])
print("Variance of Temperature:", var_temp)
```

```python
standard_dev_temp = np.std(df['Temperature (C)'])
print("Standard Deviation of Temperature:", standard_dev_temp)
```

```python
for i in range(1,13):
    month = df.loc[df['month'] == i] ["Temperature (C)"]
    print("For month",i)

    print("Mean of Temperature:", np.mean(month))
    print("Variance of Temperature:", np.var(month))
    print("Standard Deviation of Temperature:", np.std(month))

    print("--------------------------------------------------------------")
```
