```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
a = np.array([22,87,5,43,56,73,55,54,11,20,51,5,79,31,27])
fig, ax = plt.subplots(figsize = (7,5))
plt.hist(a, bins = [0,20,40,60,80,100])
```

```python
x = np.random.randn(10000)
print(x)

y = 0.8 ** x + np.random.randn(10000) + 25
print(y)

plt.hist(x,alpha=0.8,label='x',color='green')
plt.hist(y, alpha=0.6,label='y',color='red')
plt.legend(loc='upper right')
plt.show()
```

```python
blood_sugar_men = [113, 85, 90, 150, 149, 88, 93, 115, 135, 80, 77, 82, 129]

blood_sugar_women = [67, 98, 89, 120, 133, 150, 84, 69, 89, 79, 120, 112, 100]

type = [blood_sugar_men,blood_sugar_women]
colors =['g','r']
label = ['men','women']
bins = [80,100,125,150]


plt.hist(type,bins=bins,width=5,color=colors,label=label)

plt.title('Blood sugar analysis')
plt.xlabel('Blood suger level')
plt.ylabel('Frequency')
plt.legend()
plt.show()
```
