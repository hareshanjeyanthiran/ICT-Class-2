```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv('Titanic Dataset.csv')
df.info()
```

```python
df.isnull().sum()
```

```python
age_q1 = np.quantile(df['Age'], 0.25)
age_q2 = np.quantile(df['Age'], 0.5)
age_q3 = np.quantile(df['Age'], 0.75)
```

```python
print("Age Quantiles --->")
print(f"Q1: {age_q1}")
print(f"Q2: {age_q2}")
print(f"Q3: {age_q3}")
```

```python
IQR_age = age_q3 - age_q1
print(f"Interquartile Range (IQR) for Age: {IQR_age}")
```

```python
plt.hist(df['Age'])
plt.xlabel('Age')
plt.ylabel('Count of passengers')
```

```python
fare_q1 = np.quantile(df['Fare'], 0.25)
fare_q2 = np.quantile(df['Fare'], 0.5)
fare_q3 = np.quantile(df['Fare'], 0.75)
```

```python
print("Fare Quantiles --->")
print(f"Q1: {fare_q1}")
print(f"Q2: {fare_q2}")
print(f"Q3: {fare_q3}")
```

```python
IQR_fare = fare_q3 - fare_q1
print(f"Interquartile Range (IQR) for Fare: {IQR_fare}")
```

```python
bins = np.arange(0, 250, 20)
plt.hist(df['Fare'], bins=bins, edgecolor='black', alpha=0.7, color='green')
plt.xlabel('Fare')
plt.ylabel('Count of passengers')
plt.xticks(bins)
plt.show()
```

```python
plt.boxplot(df['Age'])
plt.title('Age distribution')
plt.show()
```

```python
plt.boxplot(df['Fare'])
plt.title('Fare distribution')
plt.show()
```

```python
plt.boxplot(df['Pclass'])
plt.title('Pclass distribution')
plt.show()
```
