```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import statistics as stats
```

```python
df = pd.read_csv('Titanic Dataset.csv')
df.head(5)
```

```python
mean_age = df['Age'].mean()
print(f"The mean age of passengers is: {mean_age}")
```

```python
mean_fare = df['Fare'].mean()
print(f"The mean fare of passengers is: {mean_fare}")
```

```python
median_age = np.median(df['Age'])
print(f"The median age of passengers is: {median_age}")
```

```python
median_fare = np.median(df['Fare'])
print(f"The median fare of passengers is: {median_fare}")
```

```python
mode_age = stats.mode(df['Age'])
print(f"The mode age of passengers is: {mode_age}")
```

```python
mode_class = stats.mode(df['Pclass'])
print(f"The mode class of passengers is: {mode_class}")
```

```python
num_genders = df['Gender'].value_counts()
num_genders.plot(kind='pie', autopct='%1.1f%%', startangle=90, colors=['#ff9999','#66b3ff'])
```

```python
mode_gender = df['Gender'].value_counts().index[0]
print(f"The mode gender of passengers is: {mode_gender}")
```
