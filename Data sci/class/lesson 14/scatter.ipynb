```python
import matplotlib.pyplot as plt

x = [3, 9, 5, 2, 7, 1, 4, 6, 8, 0]
y = [1, 2, 3, 4, 5, 6, 7, 8, 9, 0]

plt.scatter(x, y)
plt.xlabel('X-axis')
plt.ylabel('Y-axis')
plt.title('Scatter Plot')
plt.show()
```

```python
x1 = [99,78, 56, 45, 23, 12, 34, 67, 89, 90]
y1 = [12, 34, 56, 78, 90, 23, 45, 67, 89, 99]

x2 = [36,9,12, 45, 67, 89, 90, 23, 34, 56]
y2 = [23, 45, 67, 89, 90, 12, 34, 56, 78, 99]

plt.scatter(x1, y1, color='red', label='Group 1', marker='o', s=100)
plt.scatter(x2, y2, color='blue', label='Group 2', marker='x', s=100)
plt.xlabel('X-axis')
plt.ylabel('Y-axis')
plt.title('Scatter Plot with Two Groups')
plt.legend(loc='lower left')
plt.show()
```
