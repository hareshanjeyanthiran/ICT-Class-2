```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
data = pd.read_csv('D:/Hareshan/ICT class/Data sci/class/lesson 18/Titanic Dataset.csv')
data.info()
```

```python
data.head()
```

```python
data.isnull().sum()
```

```python
minimum_age = data['Age'].min()
print("Minimum Age:", minimum_age)
```

```python
maximum_age = data['Age'].max()
print("Maximum Age:", maximum_age)
```

```python
bins = [0, 15, 30, 45, 60, 75]
data['bined_age'] = pd.cut(data['Age'], bins)
print(data[['bined_age', 'Age']])
```

```python
age_labels = ['Young', 'Young - Adult', 'Middle Aged', 'Middle-Older Age', 'Senior']
data['bined_age'] = pd.cut(data['Age'], bins, labels=age_labels)
print(data[['bined_age', 'Age']])
```

```python
data['bined_age'].value_counts().plot(kind='bar', color='red', alpha=0.7)
plt.title('Distribution of Binned Age Categories')
plt.xlabel('Age Categories')
plt.ylabel('Count')
plt.show()
```

```python
labels = ['PassengerId','Survived','Pclass','Age','SibSp','Parch','Fare']
for i in labels:
    sns.displot(data[i])
    plt.title(f'Distribution of {i}')
    plt.xlabel(i)
    plt.ylabel('Density')
    plt.show()
```

```python
sns.boxplot(data=data, x='Embarked', y='Age')
plt.show()
```

```python
plt.scatter(x = data['Fare'], y = data['Survived'])
plt.xlabel('Fare')
plt.ylabel('Survived')
plt.title('Scatter Plot of Fare vs Survived')
plt.show()
```

```python
plt.scatter(x = data['Parch'], y = data['Survived'])
plt.xlabel('Parch')
plt.ylabel('Survived')
plt.title('Scatter Plot of Parch vs Survived')
plt.show()
```

```python
plt.scatter(x = data['SibSp'], y = data['Survived'])
plt.xlabel('SibSp')
plt.ylabel('Survived')
plt.title('Scatter Plot of SibSp vs Survived')
plt.show()
```
