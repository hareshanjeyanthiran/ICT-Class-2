```python
import pandas as pd
import numpy as np
import statistics as stats
```

```python
df = pd.read_csv('Bestsellers with categories.csv')
df.head(5)
```

```python
df.dtypes
```

```python
df.isnull().sum()
```

```python
User_Rating_mean = df['User Rating'].mean()
print("Mean of User Rating:", User_Rating_mean)
```

```python
User_Rating_median = np.median(df['User Rating'])
print("Median of User Rating:", User_Rating_median)
```

```python
User_Rating_mode = stats.mode(df['User Rating'])
print("Mode of User Rating:", User_Rating_mode)
```

```python
Price_mean = df['Price'].mean()
print("Mean of Price:", Price_mean)
```

```python
Price_median = np.median(df['Price'])
print("Median of Price:", Price_median)
```

```python
Price_mode = stats.mode(df['Price'])
print("Mode of Price:", Price_mode)
```

```python
Reviews_mean = df['Reviews'].mean()
print("Mean of Reviews:", Reviews_mean)
```

```python
Reviews_median = np.median(df['Reviews'])
print("Median of Reviews:", Reviews_median)
```

```python
Reviews_mode = stats.mode(df['Reviews'])
print("Mode of Reviews:", Reviews_mode)
```
