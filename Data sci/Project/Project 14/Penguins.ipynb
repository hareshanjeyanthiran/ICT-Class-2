```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
import seaborn as sns
```

```python
df = pd.read_csv("penguins Data.csv")
df.info()
```

```python
plt.figure(figsize=(10, 6))
plt.scatter(df['Culmen Length (mm)'], df['Body Mass (g)'], cmap='rainbow', alpha=0.7)
plt.xlabel('Culmen Length (mm)')
plt.ylabel('Body Mass (g)')
plt.title('Culmen Length vs Body Mass of Penguins')
plt.show()
```

```python
plt.figure(figsize=(10, 6))
plt.scatter(df['Culmen Depth (mm)'], df['Body Mass (g)'], cmap='rainbow',color='red' , alpha=0.7)
plt.xlabel('Culmen Depth (mm)')
plt.ylabel('Body Mass (g)')
plt.title('Culmen Depth vs Body Mass of Penguins')
plt.show()
```

```python
sns.pairplot(df, hue='Species', palette='Set2')
```

```python
numeric_df = df.select_dtypes(include='number')

plt.figure(figsize=(10, 8))
sns.heatmap(
        numeric_df.corr(),
        annot=True,
        cmap='coolwarm',
        fmt='.2f',
        linewidths=0.5
    )
plt.title('Correlation Heatmap')
plt.show()
```
