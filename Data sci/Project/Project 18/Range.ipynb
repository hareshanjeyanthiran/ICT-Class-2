```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv('Bestsellers with categories.csv')
df.info()
```

```python
df.isnull().sum()
```

```python
price_10th_percentile = df["Price"].quantile(0.10)
print(price_10th_percentile)
```

```python
price_q1 = df["Price"].quantile(0.25)
price_q2 = df["Price"].quantile(0.50)
price_q3 = df["Price"].quantile(0.75)
```

```python
print("Price Quatiles------>")
print("Q1:", price_q1)
print("Q2:", price_q2)
print("Q3:", price_q3)
```

```python
IQR_price = price_q3 - price_q1
print("Interquartile Range (IQR) for Price:", IQR_price)
```

```python
rating_q1 = df["User Rating"].quantile(0.25)
rating_q2 = df["User Rating"].quantile(0.50)
rating_q3 = df["User Rating"].quantile(0.75)
```

```python
print("User Rating Quatiles------>")
print("Q1:", rating_q1)
print("Q2:", rating_q2)
print("Q3:", rating_q3)
```

```python
IQR_rating = rating_q3 - rating_q1
print("Interquartile Range (IQR) for User Rating:", IQR_rating)
```

```python
plt.figure(figsize=(10, 6))
plt.boxplot([df["Price"], df["User Rating"]])
plt.xticks([1, 2], ["Price", "User Rating"])
plt.yticks(np.arange(0, 100, 5))
plt.title("Boxplot of Price and User Rating")
plt.ylabel("Value")
plt.show()
```
