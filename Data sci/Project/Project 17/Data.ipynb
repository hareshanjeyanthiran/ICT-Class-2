```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv('Bestsellers with categories.csv')
df.info()
```

```python
df.isnull().sum()
```

```python
print("Variance of User Rating:", df['User Rating'].var())
print("Standard Deviation of User Rating:", df['User Rating'].std())
```

```python
print("Variance of Price:", df['Price'].var())
print("Standard Deviation of Price:", df['Price'].std())
```

```python
plt.hist(df['User Rating'], bins=20)
plt.xlabel('User Rating')
plt.ylabel('Frequency')
plt.title('Distribution of User Rating')
plt.show()
```

```python
plt.hist(df['Price'], bins=20, range=(0, 50), color='orange', edgecolor='black')
plt.xlabel('Price')
plt.ylabel('Frequency')
plt.title('Distribution of Price')
plt.show()
```
