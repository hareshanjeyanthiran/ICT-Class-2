```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv('Penguins Data.csv')
df.info()
```

```python
features = [
    "Culmen Length (mm)",
    "Culmen Depth (mm)",
    "Flipper Length (mm)",
    "Body Mass (g)"
]

for feature in features:
   plt.figure(figsize=(10, 6))
   plt.hist(df[feature].dropna(), bins=30, color='blue', alpha=0.7)
   plt.xlabel(feature)
   plt.ylabel('Frequency')
   plt.title(f'Histogram of {feature}')
```

```python
plt.figure(figsize=(8, 5))

plt.hist(
    df["Culmen Length (mm)"].dropna(),
    bins=15,
    range=(30, 60),
    color="skyblue"
)

plt.xlabel("Culmen Length (mm)")
plt.ylabel("Frequency")
plt.title("Culmen Length Distribution")
plt.show()
```

```python
plt.figure(figsize=(8, 5))
plt.hist(df["Culmen Depth (mm)"].dropna(),
         bins=15, range=(12, 22), color="lightgreen")
plt.xlabel("Culmen Depth (mm)")
plt.ylabel("Frequency")
plt.title("Culmen Depth Distribution")
plt.show()
```

```python
plt.figure(figsize=(8, 5))
plt.hist(df["Flipper Length (mm)"].dropna(),
         bins=15, range=(170, 235), color="orange")
plt.xlabel("Flipper Length (mm)")
plt.ylabel("Frequency")
plt.title("Flipper Length Distribution")
plt.show()
```

```python
plt.figure(figsize=(8, 5))
plt.hist(df["Body Mass (g)"].dropna(),
         bins=15, range=(2500, 6500), color="plum")
plt.xlabel("Body Mass (g)")
plt.ylabel("Frequency")
plt.title("Body Mass Distribution")
plt.show()
```
